```python
import zipfile
import os

dataset = '/content/dl-circuit-dataset-bp.zip'

with zipfile.ZipFile(dataset, 'r') as zip_ref:
    zip_ref.extractall()
```

```python
from ultralytics import YOLO

model = YOLO("yolo11m-obb.pt")

results = model.train(
    data = "/content/data.yaml",
    batch = 16,
    imgsz = 640,
    pretrained = True,
    epochs = 200,
    optimizer = "AdamW",
    lr0 = 0.001,
    cos_lr = True,
    patience = 200,

)
```

#Test Circuit 1 (Notebook)

```python
import cv2
import matplotlib.pyplot as plt
from ultralytics import YOLO

model = YOLO("/content/runs/obb/train/weights/last.pt")

image = "/content/Circuit_Dataset_YOLO/Circuit_Dataset_YOLO/Testphotos/Circuit_Test_1_Notebook.png"


inference = model.predict(source=image, save=True, show=False)


for result in inference:

    im_bgr = result.plot()


    im_plt = cv2.cvtColor(im_bgr, cv2.COLOR_BGR2RGB)


    plt.figure(figsize=(8, 8))
    plt.imshow(im_plt)
    plt.axis('off')
    plt.show()
```
